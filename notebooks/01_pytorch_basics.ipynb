```python
import platform
import torch

print('Python:', platform.python_version())
print('PyTorch:', torch.__version__)
print('MPS available:', torch.backends.mps.is_available())
```

## Steps

### 1. 创建张量

张量（tensor）可以先理解为 PyTorch 中更适合计算的数字列表。

```python
numbers = torch.tensor([1, 2, 3])
print(numbers)
print('Shape:', numbers.shape)
```

### 2. 张量运算

试着把下面的 `* 2` 改成 `+ 10` 后重新运行，观察结果。

```python
doubled = numbers + 20
print(doubled)
```

### 3. 自动求导

`requires_grad=True` 表示让 PyTorch 追踪这个变量的计算过程，以便计算导数。

```python
x = torch.tensor(3.0, requires_grad=True)
y = x ** 3 + 2 * x * x
y.backward()

print('y =', y.item())
print('dy/dx =', x.grad.item())
```

- `MPS available` 为 `True` 表示你的 Mac 可以使用 Apple GPU 加速。
- 自动求导示例中，`x = 3` 时 `y = 15`，导数应为 `8`。
